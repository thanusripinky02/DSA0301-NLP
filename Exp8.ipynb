```python
import random

words = ["I", "like", "dogs", "run"]

# Basic probability model
probabilities = {
    "I": ["PRP"],
    "like": ["VB", "IN"],
    "dogs": ["NNS"],
    "run": ["VB", "NN"]
}

for word in words:
    tag = random.choice(probabilities[word])
    print(word, "->", tag)
```
