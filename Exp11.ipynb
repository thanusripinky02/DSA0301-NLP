```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V", "NP"]],
    "Det": [["the"]],
    "N": [["cat"], ["dog"]],
    "V": [["sees"]]
}

def parse(symbols, words, pos=0):
    if not symbols:
        return pos == len(words)

    symbol = symbols[0]

    if symbol not in grammar:
        return pos < len(words) and words[pos] == symbol and \
               parse(symbols[1:], words, pos + 1)

    for rule in grammar[symbol]:
        if parse(rule + symbols[1:], words, pos):
            return True

    return False

sentence = "the cat sees the dog".split()

if parse(["S"], sentence):
    print("Sentence is accepted")
else:
    print("Sentence is rejected")
```
