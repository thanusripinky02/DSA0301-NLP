```python
from nltk.stem import PorterStemmer

# Create Porter Stemmer object
ps = PorterStemmer()

# List of words
words = ["playing", "played", "plays", "studies", "studying",
         "connected", "connection", "running", "easily"]

# Perform stemming
print("Original Word -> Stemmed Word")

for word in words:
    print(word, "->", ps.stem(word))
```
