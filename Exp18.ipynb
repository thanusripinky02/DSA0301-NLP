```python
import re

expression = "P(x) AND Q(x)"

pattern = r'([A-Z])\((\w+)\)\s+(AND|OR)\s+([A-Z])\((\w+)\)'

match = re.match(pattern, expression)

if match:
    p1, arg1, operator, p2, arg2 = match.groups()

    print("Predicate 1:", p1)
    print("Argument 1:", arg1)
    print("Operator:", operator)
    print("Predicate 2:", p2)
    print("Argument 2:", arg2)
else:
    print("Invalid FOPC expression")
```
