```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V", "NP"]],
    "Det": [["the"]],
    "N": [["cat"], ["dog"]],
    "V": [["sees"]]
}

sentence = "the cat sees the dog".split()
index = 0

def parse(symbol):
    global index

    if symbol not in grammar:
        if index < len(sentence) and sentence[index] == symbol:
            index += 1
            return (symbol, [])
        return None

    for rule in grammar[symbol]:
        old_index = index
        children = []

        for item in rule:
            child = parse(item)
            if child:
                children.append(child)
            else:
                index = old_index
                break
        else:
            return (symbol, children)

    return None

def print_tree(tree, level=0):
    print("  " * level + tree[0])
    for child in tree[1]:
        print_tree(child, level + 1)

tree = parse("S")

if tree and index == len(sentence):
    print_tree(tree)
else:
    print("No parse tree found")
```
