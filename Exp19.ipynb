```python
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

sentence = "I went to the bank to deposit money."

words = word_tokenize(sentence)

sense = lesk(words, "bank")

print("Word:", "bank")
print("Meaning:", sense.definition())
```
