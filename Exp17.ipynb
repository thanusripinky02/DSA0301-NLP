```python
from nltk.corpus import wordnet

word = "bank"

synsets = wordnet.synsets(word)

print("Synsets for:", word)

for syn in synsets[:5]:
    print("\nSynset:", syn.name())
    print("Definition:", syn.definition())
    print("Examples:", syn.examples())
```
