```python
import re

text = "John went to the park. He played football. Mary saw him."

names = ["John", "Mary"]

for sentence in text.split("."):
    sentence = sentence.strip()

    if "He" in sentence:
        print(sentence, "-> He refers to John")

    elif "him" in sentence:
        print(sentence, "-> him refers to John")

    elif "She" in sentence:
        print(sentence, "-> She refers to Mary")
```
