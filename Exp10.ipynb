```python
words = ["The", "boy", "is", "playing", "football"]

# Initial tags
tags = ["NN", "NN", "NN", "NN", "NN"]

# Transformation rules

for i in range(len(words)):
    if words[i].lower() in ["the", "a", "an"]:
        tags[i] = "DT"

    elif words[i].lower() in ["is", "am", "are", "was", "were"]:
        tags[i] = "VB"

    elif words[i].endswith("ing"):
        tags[i] = "VBG"

print("Word -> POS Tag")

for word, tag in zip(words, tags):
    print(word, "->", tag)
```
