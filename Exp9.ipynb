```python
import re

text = "The cats are running quickly."

words = text.split()

for word in words:
    word = word.strip(".,!?")

    if re.match(r'.*ing$', word):
        tag = "VBG"
    elif re.match(r'.*ly$', word):
        tag = "RB"
    elif re.match(r'.*s$', word):
        tag = "NNS"
    elif word.lower() in ["the", "a", "an"]:
        tag = "DT"
    else:
        tag = "NN"

    print(word, "->", tag)
```
