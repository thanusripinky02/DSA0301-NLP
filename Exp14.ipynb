```python
grammar = {
    "singular": ["cat", "dog", "boy", "girl"],
    "plural": ["cats", "dogs", "boys", "girls"]
}

def check_agreement(sentence):
    words = sentence.split()

    if len(words) != 3:
        return False

    subject = words[0]
    verb = words[1]

    if subject in grammar["singular"] and verb == "runs":
        return True

    if subject in grammar["plural"] and verb == "run":
        return True

    return False

sentences = [
    "dog runs",
    "dogs run"
]

for sentence in sentences:
    print(sentence, "->",
          "Correct agreement" if check_agreement(sentence)
          else "Incorrect agreement")
```
