```python
import re

text = """
Python is a programming language.
Python is easy to learn.
Programming languages are used to create software.
"""

sentences = [s.strip() for s in text.split(".") if s.strip()]

print("Number of sentences:", len(sentences))

# Calculate word overlap between consecutive sentences
for i in range(len(sentences) - 1):
    words1 = set(re.findall(r'\w+', sentences[i].lower()))
    words2 = set(re.findall(r'\w+', sentences[i + 1].lower()))

    overlap = words1.intersection(words2)

    print("Sentence", i + 1, "and", i + 2)
    print("Common words:", overlap)

if len(sentences) > 1:
    print("Coherence: Good")
else:
    print("Coherence: Poor")
```
