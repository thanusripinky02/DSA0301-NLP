```python
import re

text = "My email is student123@gmail.com and my phone number is 9876543210."

# Search for an email pattern
email = re.search(r'\b[\w.-]+@[\w.-]+\.\w+\b', text)

# Search for a phone number pattern
phone = re.search(r'\b\d{10}\b', text)

# Display results
if email:
    print("Email found:", email.group())

if phone:
    print("Phone number found:", phone.group())

# Match a word pattern
words = re.findall(r'\bstudent\w*', text)
print("Matching words:", words)
```
