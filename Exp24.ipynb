```python
def dialog_act(sentence):
    sentence = sentence.lower()

    if sentence.endswith("?"):
        return "Question"

    elif any(word in sentence for word in ["hello", "hi", "good morning"]):
        return "Greeting"

    elif any(word in sentence for word in ["thank you", "thanks"]):
        return "Thanking"

    elif any(word in sentence for word in ["please", "could you", "can you"]):
        return "Request"

    elif any(word in sentence for word in ["bye", "goodbye"]):
        return "Closing"

    else:
        return "Statement"


dialog = [
    "Hello!",
    "Can you help me?",
    "Thank you!",
    "Goodbye!"
]

for sentence in dialog:
    print(sentence, "->", dialog_act(sentence))
```
