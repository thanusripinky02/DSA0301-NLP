```python
import spacy

nlp = spacy.load("en_core_web_sm")

sentence = "The student reads a book."

doc = nlp(sentence)

print("Noun Phrases and their meanings:")

for chunk in doc.noun_chunks:
    print("Noun Phrase:", chunk.text)

    root = chunk.root
    print("Meaning/Role:", root.dep_)
    print("Head Word:", root.text)
    print()
```
